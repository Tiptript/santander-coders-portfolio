```python
import json

def lerArquivoUsuarios(nomeArquivo='projetoModuloII.json'):
    try:
        with open(nomeArquivo, 'r', encoding='utf-8') as arquivo:
            usuarios = json.load(arquivo)
            return usuarios
    except FileNotFoundError:
        return {}

def salvarArquivoUsuarios(usuarios, nomeArquivo='projetoModuloII.json'):
    with open(nomeArquivo, 'w', encoding='utf-8') as arquivo:
        json.dump(usuarios, arquivo)

def inserirUsuario(usuarios, nome, telefone='vazio', endereco='vazio'):
    for id, usuario in usuarios.items():
        if usuario['Nome'] == nome and usuario['Telefone'] == telefone and usuario['Endereço'] == endereco:
            if not usuario['Status']:
                usuario['Status'] = True
                print(f"Usuário com ID {id} reativado.")
            else:
                print("Usuário já cadastrado.")
            return

    novo_id = str(int(max(usuarios.keys(), default=0)) + 1)
    usuarios[novo_id] = {
        'Nome': nome,
        'Telefone': telefone,
        'Endereço': endereco,
        'Status': True
    }
    print(f"Usuário com ID {novo_id} inserido.")

def excluirUsuario(usuarios, *ids):
    for id in ids:
        if id in usuarios:
            usuarios[id]['Status'] = False
            print(f"Usuário com ID {id} excluído.")
        else:
            print(f"Usuário com ID {id} não encontrado.")

def atualizarUsuario(usuarios, id):
    if id in usuarios:
        usuario = usuarios[id]
        print(f"Informações do usuário com ID {id}:")
        while True:
            print(f"1 - Nome: {usuario['Nome']}")
            print(f"2 - Telefone: {usuario['Telefone']}")
            print(f"3 - Endereço: {usuario['Endereço']}")
            print("0 - Sair")

            opcao = input("Qual informação deseja alterar? (Digite o número da opção): ")
            if opcao == '1':
                novo_nome = input("Insira o novo nome: ")
                usuario['Nome'] = novo_nome
            elif opcao == '2':
                novo_telefone = input("Insira o novo telefone: ")
                usuario['Telefone'] = novo_telefone
            elif opcao == '3':
                novo_endereco = input("Insira o novo endereço: ")
                usuario['Endereço'] = novo_endereco
            elif opcao == '0':
                print("Atualização concluída.")
                break
            else:
                print("Opção inválida. Tente novamente.")
    else:
        print(f"Usuário com ID {id} não encontrado!")

def exibirUmUsuario(usuarios, id):
    if id in usuarios:
        usuario = usuarios[id]
        print(f"Informações do usuário com ID {id}:")
        print(f"Nome: {usuario['Nome']}")
        print(f"Telefone: {usuario['Telefone']}")
        print(f"Endereço: {usuario['Endereço']}")
    else:
        print(f"Usuário com ID {id} não encontrado!")

def exibirTodosUsuarios(usuarios):
    print("\nLista de usuários ativos:")
    for id, usuario in usuarios.items():
        if usuario.get('Status', True):
            print(f"ID: {id}")
            print(f"Nome: {usuario['Nome']}")
            print(f"Telefone: {usuario['Telefone']}")
            print(f"Endereço: {usuario['Endereço']}")
            print()

def main():
    usuarios = lerArquivoUsuarios()
    while True:
        print("\nBoas vindas ao nosso sistema:")
        print("1 - Inserir usuário")
        print("2 - Excluir usuário")
        print("3 - Atualizar usuário")
        print("4 - Informações de um usuário")
        print("5 - Informações de todos os usuários")
        print("6 - Sair")
        
        opcao = input("Escolha uma opção: ")

        if opcao == '1':
            nome = input("Insira o nome do usuário: ")
            telefone = input("Insira o telefone do usuário: ")
            endereco = input("Insira o endereço do usuário: ")
            inserirUsuario(usuarios, nome, telefone, endereco)
        elif opcao == '2':
            ids_excluir = input("Insira o ID para excluir: ").split(',')
            excluirUsuario(usuarios, *ids_excluir)
        elif opcao == '3':
            id_atualizar = input("Insira o ID para atualizar: ")
            atualizarUsuario(usuarios, id_atualizar)
        elif opcao == '4':
            id_exibir = input("Insira o ID para exibir: ")
            exibirUmUsuario(usuarios, id_exibir)
        elif opcao == '5':
            exibirTodosUsuarios(usuarios)
        elif opcao == '6':
            salvarArquivoUsuarios(usuarios)
            print("Saindo.")
            break
        else:
            print("Opção inválida. Tente novamente.")

if __name__ == "__main__":

    main()
```
